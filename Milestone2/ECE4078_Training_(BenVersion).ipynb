```python
# Import requirements
from ultralytics import YOLO, utils
# prevent ultralytics from tracking your activity
!yolo settings sync=False

from IPython.display import display, Image
import cv2
import matplotlib
```

```python
%cd /content/ECE4078Fruits_datasetv1_yolov11/
```

```python
# Load the trained YOLOv11 Small model
model = YOLO("yolo11s.pt")
# OR load the trained our trained model
model = YOLO("/content/drive/MyDrive/ECE4078_B1/Models/InitialModelV1/ECE4078B1_Modelv1_FixedNames.pt")
try:
  img = cv2.imread('/content/ECE4078Fruits_datasetv1_yolov11/test/images/capsicum_18_png.rf.ee73d32d372a5f4522d8c852df7a2f13.jpg')
  # Get dimensions (height, width, channels)
  height, width, channels = img.shape
  print(f"Image dimensions: {width}x{height}")
except:
  pass
```

```python
utils.checks.check_yolo(verbose=True, device='')
```

```python
# Train the model
results = model.train(
    data="/content/ECE4078Fruits_datasetv1_yolov11/data.yaml",
    epochs=60, # Train for 60 epochs
    imgsz=320, # Length of longest side
    rect=True, # Enable rectangular images
    plots=True # Enable plots
    )
```

```python
# Run Prediction
results = model.val(data="data.yaml", split="test")
```
